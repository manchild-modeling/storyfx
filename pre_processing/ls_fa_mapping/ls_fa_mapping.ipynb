```python
from pathlib import Path
import json
import re

EXCERPT_ID = "excerpt_000"

# ls_fa_mapping.ipynb.ipynb:
# storyfx/pre_processing/ls_fa_mapping/ls_fa_mapping.ipynb

PROJECT_ROOT = Path.cwd().parents[1]

EXCERPT_DIR = PROJECT_ROOT / "data" / EXCERPT_ID

LABELSTUDIO_DIR = EXCERPT_DIR / "label_studio"
ALIGNMENT_DIR = EXCERPT_DIR / "alignment"
ANNOTATIONS_DIR = EXCERPT_DIR / "annotations"

ANNOTATIONS_DIR.mkdir(parents=True, exist_ok=True)


# ============================================================
# INPUTS
# ============================================================

FA_FILE = (
    ALIGNMENT_DIR /
    f"{EXCERPT_ID}_fa_parsed.txt"
)

SEGMENTS_LS_FILE = (
    LABELSTUDIO_DIR /
    f"{EXCERPT_ID}_segments_ls.json"
)

DISCRETE_SFX_LS_FILE = (
    LABELSTUDIO_DIR /
    f"{EXCERPT_ID}_discfx_ls.json"
)

AMBIENT_SFX_LS_FILE = (
    LABELSTUDIO_DIR /
    f"{EXCERPT_ID}_ambfx_ls.json"
)


# ============================================================
# OUTPUTS
# ============================================================

SEGMENTS_FINAL_FILE = (
    ANNOTATIONS_DIR /
    f"{EXCERPT_ID}_segments_final.json"
)

DISCRETE_SFX_FINAL_FILE = (
    ANNOTATIONS_DIR /
    f"{EXCERPT_ID}_discrete_sfx_final.json"
)

AMBIENT_SFX_FINAL_FILE = (
    ANNOTATIONS_DIR /
    f"{EXCERPT_ID}_ambient_sfx_final.json"
)


print("Project root:", PROJECT_ROOT)
print("Excerpt:", EXCERPT_ID)

print("\nFA:")
print(FA_FILE)

print("\nLabel Studio:")
print(SEGMENTS_LS_FILE)
print(DISCRETE_SFX_LS_FILE)
print(AMBIENT_SFX_LS_FILE)

print("\nFinal annotations:")
print(SEGMENTS_FINAL_FILE)
print(DISCRETE_SFX_FINAL_FILE)
print(AMBIENT_SFX_FINAL_FILE)
```

```python
def clean_text(text):
    """
    Normalize text for comparison with MFA tokens.
    """

    if not text:
        return ""

    if isinstance(text, list):
        text = " ".join(text)

    text = text.replace("\\n", " ").replace("\n", " ")

    # MFA may separate compounds
    text = text.replace("-", " ")

    # Lowercase + remove punctuation
    text = re.sub(
        r"[^\w\s]",
        "",
        text.lower()
    )

    return " ".join(text.split())


def load_fa_words(fa_path):
    """
    Load parsed MFA representation:

        word|0.42 another|0.81 ...

    Returns ordered words with their MFA end timestamps.
    """

    with open(fa_path, "r", encoding="utf-8") as f:
        content = f.read().strip()

    fa_words = []

    for token in content.split():

        word, timestamp = token.rsplit("|", 1)

        cleaned_word = clean_text(word)

        if cleaned_word:
            fa_words.append({
                "word": cleaned_word,
                "end": float(timestamp)
            })

    return fa_words


fa_words = load_fa_words(FA_FILE)

print(f"Loaded {len(fa_words)} MFA words.")
print(fa_words[:10])
```

```python
from difflib import SequenceMatcher


def find_text_span(
    region_text,
    fa_words,
    search_start_idx=0,
    anchor_size=5,
    fuzzy_threshold=0.80
):
    """
    Resolve a Label Studio text span against MFA words.

    Strategy:
    1. Exact head/tail anchor matching.
    2. If exact matching fails, use fuzzy token matching.
    3. Preserve chronological search order where possible.

    Returns:
        {
            "start_anchor_time": float,
            "end_anchor_time": float,
            "start_word_index": int,
            "end_word_index": int,
            "next_search_index": int
        }
    """

    target_words = clean_text(region_text).split()

    if not target_words:
        return None

    # --------------------------------------------------------
    # Phrase matcher
    # --------------------------------------------------------

    def find_phrase(
        phrase,
        start_idx,
        allow_fuzzy=True
    ):
        n = len(phrase)

        # ================================================
        # 1. EXACT MATCH
        # ================================================

        for i in range(
            start_idx,
            len(fa_words) - n + 1
        ):

            candidate = [
                fa_words[i + j]["word"]
                for j in range(n)
            ]

            if candidate == phrase:
                return i, i + n - 1


        # ================================================
        # 2. FUZZY MATCH
        # ================================================

        if not allow_fuzzy:
            return None

        best_match = None
        best_score = 0.0

        # Allow MFA sequence to be slightly shorter/longer
        # than the source phrase.
        min_window = max(1, n - 2)
        max_window = n + 2

        target_string = " ".join(phrase)

        for window_size in range(
            min_window,
            max_window + 1
        ):

            for i in range(
                start_idx,
                len(fa_words) - window_size + 1
            ):

                candidate_words = [
                    fa_words[i + j]["word"]
                    for j in range(window_size)
                ]

                candidate_string = " ".join(
                    candidate_words
                )

                score = SequenceMatcher(
                    None,
                    target_string,
                    candidate_string
                ).ratio()

                if score > best_score:

                    best_score = score

                    best_match = (
                        i,
                        i + window_size - 1
                    )


        if (
            best_match is not None
            and best_score >= fuzzy_threshold
        ):

            print(
                f"⚠️ Fuzzy FA match "
                f"({best_score:.2f}): "
                f"'{' '.join(phrase)}'"
            )

            return best_match


        return None


    # ========================================================
    # SHORT SEGMENTS
    # ========================================================

    # For small segments it is safer to match the whole thing.
    if len(target_words) <= 10:

        match = find_phrase(
            target_words,
            search_start_idx
        )

        # Global fallback
        if match is None and search_start_idx > 0:

            match = find_phrase(
                target_words,
                0
            )

        if match is None:
            return None

        start_idx, end_idx = match


    # ========================================================
    # LONG SEGMENTS
    # ========================================================

    else:

        k = min(
            anchor_size,
            len(target_words)
        )

        head = target_words[:k]
        tail = target_words[-k:]


        # ----------------------------
        # Find head
        # ----------------------------

        head_match = find_phrase(
            head,
            search_start_idx
        )

        if (
            head_match is None
            and search_start_idx > 0
        ):

            head_match = find_phrase(
                head,
                0
            )


        if head_match is None:
            return None


        start_idx = head_match[0]


        # ----------------------------
        # Find tail
        # ----------------------------

        tail_search_start = (
            head_match[1] + 1
        )

        tail_match = find_phrase(
            tail,
            tail_search_start
        )


        if tail_match is None:
            return None


        end_idx = tail_match[1]


    # ========================================================
    # RESULT
    # ========================================================

    return {
        "start_anchor_time":
            fa_words[start_idx]["end"],

        "end_anchor_time":
            fa_words[end_idx]["end"],

        "start_word_index":
            start_idx,

        "end_word_index":
            end_idx,

        "next_search_index":
            end_idx + 1
    }
```

```python
def quantify_segments(
    labelstudio_path,
    fa_words,
    excerpt_id
):
    """
    Convert Label Studio segment annotations into canonical
    StoryFX segment annotations with MFA time anchors.
    """

    with open(
        labelstudio_path,
        "r",
        encoding="utf-8"
    ) as f:
        ls_data = json.load(f)

    raw_segments = []


    # ========================================================
    # EXTRACT LABEL STUDIO REGIONS
    # ========================================================

    for item in ls_data:

        for annotation in item.get("annotations", []):

            regions = {}

            for result in annotation.get("result", []):

                region_id = result.get("id")

                if not region_id:
                    continue

                if region_id not in regions:
                    regions[region_id] = {
                        "labelstudio_id": region_id,
                        "text": "",
                        "labels": [],
                        "choices": []
                    }

                value = result.get("value", {})
                from_name = result.get("from_name")


                # Text span
                if (
                    "text" in value
                    and not isinstance(value["text"], list)
                ):
                    regions[region_id]["text"] = (
                        value["text"]
                    )


                # Segment type
                #
                # Supports both your newer terminology and
                # legacy Label Studio configurations.
                if from_name in {"segment", "scene"}:
                    regions[region_id]["labels"].extend(
                        value.get("labels", [])
                    )


                # Reverb profile
                elif from_name in {
                    "reverb_profile",
                    "spatial_profile"
                }:
                    choices = value.get("choices", [])

                    if choices:
                        regions[region_id]["choices"] = (
                            choices
                        )


            raw_segments.extend(
                region
                for region in regions.values()
                if region["text"]
            )


    # ========================================================
    # QUANTIFY IN CHRONOLOGICAL ORDER
    # ========================================================

    quantified = []

    current_search_index = 0

    for region in raw_segments:

        match = find_text_span(
            region["text"],
            fa_words,
            current_search_index
        )

        if match is None:
            raise ValueError(
                "Could not align segment:\n"
                f"{region['text'][:100]}"
            )

        current_search_index = (
            match["next_search_index"]
        )

        quantified.append({
            "labelstudio_id":
                region["labelstudio_id"],

            "segment_type":
                region["labels"][0]
                if region["labels"]
                else None,

            "start_anchor_time":
                match["start_anchor_time"],

            "end_anchor_time":
                match["end_anchor_time"],

            "reverb_profile":
                region["choices"][0]
                if region["choices"]
                else None
        })


    # ========================================================
    # CREATE CANONICAL STORYFX IDS
    # ========================================================

    for i, segment in enumerate(quantified):

        segment["segment_id"] = (
            f"{excerpt_id}_segment_{i:03d}"
        )


    # Put primary ID first just for readability
    quantified = [
        {
            "segment_id":
                segment["segment_id"],

            "labelstudio_id":
                segment["labelstudio_id"],

            "segment_type":
                segment["segment_type"],

            "start_anchor_time":
                segment["start_anchor_time"],

            "end_anchor_time":
                segment["end_anchor_time"],

            "reverb_profile":
                segment["reverb_profile"]
        }

        for segment in quantified
    ]


    return quantified
```

```python
segments_final = quantify_segments(
    SEGMENTS_LS_FILE,
    fa_words,
    EXCERPT_ID
)

with open(
    SEGMENTS_FINAL_FILE,
    "w",
    encoding="utf-8"
) as f:

    json.dump(
        segments_final,
        f,
        indent=2,
        ensure_ascii=False
    )


print(
    f"Saved {len(segments_final)} segments."
)

print(
    json.dumps(
        segments_final,
        indent=2
    )
)
```

```python
segment_id_map = {
    segment["labelstudio_id"]:
        segment["segment_id"]

    for segment in segments_final
}

print(
    json.dumps(
        segment_id_map,
        indent=2
    )
)
```

```python
def find_parent_segment(
    sfx_span,
    segments,
    overlap_threshold=0.70
):
    """
    Find the narrative segment containing a discrete-SFX
    textual anchor.

    Falls back to the segment containing the largest
    proportional overlap.
    """

    if not sfx_span:
        return None

    sfx_start = sfx_span["start"]
    sfx_end = sfx_span["end"]

    duration = sfx_end - sfx_start

    if duration < 0:
        return None


    # ========================================================
    # STRICT CONTAINMENT
    # ========================================================

    for segment in segments:

        segment_start = (
            segment["start_anchor_time"]
        )

        segment_end = (
            segment["end_anchor_time"]
        )

        if (
            sfx_start >= segment_start
            and sfx_end <= segment_end
        ):
            return segment["segment_id"]


    # ========================================================
    # OVERLAP FALLBACK
    # ========================================================

    if duration == 0:
        return None

    best_segment_id = None
    best_overlap = 0.0

    for segment in segments:

        overlap_start = max(
            sfx_start,
            segment["start_anchor_time"]
        )

        overlap_end = min(
            sfx_end,
            segment["end_anchor_time"]
        )

        overlap_duration = max(
            0,
            overlap_end - overlap_start
        )

        ratio = (
            overlap_duration / duration
        )

        if ratio > best_overlap:
            best_overlap = ratio
            best_segment_id = (
                segment["segment_id"]
            )


    if best_overlap >= overlap_threshold:

        print(
            "⚠️ Overlap fallback: "
            f"{best_segment_id} "
            f"({best_overlap:.1%})"
        )

        return best_segment_id


    return None
```

```python
def quantify_discrete_sfx(
    labelstudio_path,
    fa_words,
    segments,
    excerpt_id
):
    """
    Convert Label Studio discrete-SFX annotations into
    canonical StoryFX annotations with FA time anchors.
    """

    with open(
        labelstudio_path,
        "r",
        encoding="utf-8"
    ) as f:
        ls_data = json.load(f)


    raw_regions = []


    # ========================================================
    # EXTRACT LABEL STUDIO REGIONS
    # ========================================================

    for item in ls_data:

        for annotation in item.get(
            "annotations",
            []
        ):

            regions = {}

            for result in annotation.get(
                "result",
                []
            ):

                region_id = result.get("id")

                if not region_id:
                    continue

                if region_id not in regions:

                    regions[region_id] = {
                        "labelstudio_id":
                            region_id,

                        "text": "",

                        "caption": None,

                        "duration_category":
                            None,

                        "transmission_profile":
                            None
                    }


                value = result.get(
                    "value",
                    {}
                )

                from_name = result.get(
                    "from_name"
                )


                # Selected narrative text
                if (
                    "text" in value
                    and not isinstance(
                        value["text"],
                        list
                    )
                ):
                    regions[
                        region_id
                    ]["text"] = value["text"]


                # Supports old + new LS field names
                if from_name in {
                    "discrete_sfx_region",
                    "hard_region"
                }:
                    pass


                elif (
                    from_name
                    == "duration_category"
                ):
                    choices = value.get(
                        "choices",
                        []
                    )

                    if choices:
                        regions[
                            region_id
                        ]["duration_category"] = (
                            choices[0]
                        )


                elif from_name in {
                    "transmission_profile",
                    "special_transmission_profile"
                }:

                    choices = value.get(
                        "choices",
                        []
                    )

                    if choices:
                        regions[
                            region_id
                        ][
                            "transmission_profile"
                        ] = choices[0]


                elif from_name == "caption":

                    text = value.get(
                        "text",
                        []
                    )

                    if text:

                        regions[
                            region_id
                        ]["caption"] = (
                            text[0]
                            if isinstance(
                                text,
                                list
                            )
                            else text
                        )


            raw_regions.extend(
                region
                for region in regions.values()
                if region["text"]
            )


    # ========================================================
    # TEMPORAL QUANTIFICATION
    # ========================================================

    final = []

    current_search_index = 0


    for i, region in enumerate(raw_regions):

        match = find_text_span(
            region["text"],
            fa_words,
            current_search_index
        )


        if match is None:
            raise ValueError(
                "Could not align discrete SFX "
                "annotation:\n"
                f"{region['text']}"
            )


        current_search_index = (
            match["next_search_index"]
        )


        anchor_span = {
            "start":
                match["start_anchor_time"],

            "end":
                match["end_anchor_time"]
        }


        parent_segment_id = (
            find_parent_segment(
                anchor_span,
                segments
            )
        )


        final.append({

            "discrete_sfx_id":
                f"{excerpt_id}_"
                f"discrete_sfx_{i:03d}",

            "labelstudio_id":
                region["labelstudio_id"],

            "anchor_time_span": [
                match["start_anchor_time"],
                match["end_anchor_time"]
            ],

            "segment_id":
                parent_segment_id,

            "caption":
                region["caption"],

            "duration_category":
                region[
                    "duration_category"
                ],

            "transmission_profile":
                region[
                    "transmission_profile"
                ]
        })


    return final
```

```python
discrete_sfx_final = quantify_discrete_sfx(
    DISCRETE_SFX_LS_FILE,
    fa_words,
    segments_final,
    EXCERPT_ID
)


with open(
    DISCRETE_SFX_FINAL_FILE,
    "w",
    encoding="utf-8"
) as f:

    json.dump(
        discrete_sfx_final,
        f,
        indent=2,
        ensure_ascii=False
    )


print(
    f"Saved {len(discrete_sfx_final)} "
    "discrete SFX annotations."
)

print(
    json.dumps(
        discrete_sfx_final,
        indent=2
    )
)
```

```python
def clean_ambient_sfx(
    labelstudio_path,
    segment_id_map
):
    """
    Extract ambient-SFX layer annotations and replace
    Label Studio segment IDs with canonical StoryFX IDs.
    """

    with open(
        labelstudio_path,
        "r",
        encoding="utf-8"
    ) as f:
        ls_data = json.load(f)


    regions = {}
    ordered_segment_ids = []


    for item in ls_data:

        for annotation in item.get(
            "annotations",
            []
        ):

            for result in annotation.get(
                "result",
                []
            ):

                ls_segment_id = (
                    result.get("id")
                )

                if not ls_segment_id:
                    continue


                if ls_segment_id not in regions:

                    regions[ls_segment_id] = {
                        "place_layer_caption":
                            None,

                        "weather_layer_caption":
                            None,

                        "special_layer_caption":
                            None
                    }


                value = result.get(
                    "value",
                    {}
                )

                from_name = result.get(
                    "from_name"
                )


                # Segment marked as having ambience
                #
                # Supports your old LS name as an alias.
                if (
                    from_name
                    in {"segment", "scene"}
                    and "spatial_extended"
                    in value.get(
                        "labels",
                        []
                    )
                ):

                    if (
                        ls_segment_id
                        not in ordered_segment_ids
                    ):
                        ordered_segment_ids.append(
                            ls_segment_id
                        )


                elif (
                    from_name
                    == "place_layer_caption"
                ):

                    text = value.get(
                        "text"
                    )

                    if text:
                        regions[
                            ls_segment_id
                        ][
                            "place_layer_caption"
                        ] = (
                            text[0]
                            if isinstance(
                                text,
                                list
                            )
                            else text
                        )


                elif from_name in {
                    "weather_layer_caption",
                    "weather_layer_prompt"
                }:

                    text = value.get(
                        "text"
                    )

                    if text:
                        regions[
                            ls_segment_id
                        ][
                            "weather_layer_caption"
                        ] = (
                            text[0]
                            if isinstance(
                                text,
                                list
                            )
                            else text
                        )


                elif (
                    from_name
                    == "special_layer_caption"
                ):

                    text = value.get(
                        "text"
                    )

                    if text:
                        regions[
                            ls_segment_id
                        ][
                            "special_layer_caption"
                        ] = (
                            text[0]
                            if isinstance(
                                text,
                                list
                            )
                            else text
                        )


    # ========================================================
    # CONVERT LS IDS → STORYFX IDS
    # ========================================================

    final = []


    for ls_segment_id in ordered_segment_ids:

        segment_id = segment_id_map.get(
            ls_segment_id
        )


        if segment_id is None:

            raise ValueError(
                "Ambient annotation references "
                "unknown Label Studio segment ID: "
                f"{ls_segment_id}"
            )


        final.append({

            "segment_id":
                segment_id,

            "labelstudio_segment_id":
                ls_segment_id,

            **regions[ls_segment_id]
        })


    return final
```

```python
ambient_sfx_final = clean_ambient_sfx(
    AMBIENT_SFX_LS_FILE,
    segment_id_map
)


with open(
    AMBIENT_SFX_FINAL_FILE,
    "w",
    encoding="utf-8"
) as f:

    json.dump(
        ambient_sfx_final,
        f,
        indent=2,
        ensure_ascii=False
    )


print(
    f"Saved {len(ambient_sfx_final)} "
    "ambient SFX annotations."
)

print(
    json.dumps(
        ambient_sfx_final,
        indent=2
    )
)
```
