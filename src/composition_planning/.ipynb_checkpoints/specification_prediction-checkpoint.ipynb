```python
import json
from pathlib import Path

import torch
from transformers import AutoTokenizer, AutoModelForCausalLM
```

```python
MODEL_PATH = "/workspace/models/Qwen3.8-27B"

tokenizer = AutoTokenizer.from_pretrained(
    MODEL_PATH,
    local_files_only=True,
)

model = AutoModelForCausalLM.from_pretrained(
    MODEL_PATH,
    torch_dtype=torch.bfloat16,
    device_map="auto",
    local_files_only=True,
)

model.eval()
```

```python
PROJECT_ROOT = Path("/workspace/storyfx")

excerpt = "excerpt_000"
task = "segment_specification"
strategy = "storyfx"

ENABLE_THINKING = True

MAX_NEW_TOKENS = 3000
```

```python
prompt_path = (
    PROJECT_ROOT
    / "configs"
    / "prompts"
    / strategy
    / f"{task}.txt"
)

excerpt_path = (
    PROJECT_ROOT
    / "data"
    / excerpt
    / "raw"
    / f"{excerpt}.txt"
)

output_path = (
    PROJECT_ROOT
    / "outputs"
    / "composition_plans"
    / f"pred_{excerpt}_{task}.json"
)
```

```python
prompt_template = prompt_path.read_text(
    encoding="utf-8"
).strip()

excerpt_text = excerpt_path.read_text(
    encoding="utf-8"
).strip()

prompt = (
    f"{prompt_template}\n\n"
    f"### INPUT NARRATIVE\n\n"
    f"{excerpt_text}"
)

messages = [
    {
        "role": "user",
        "content": prompt,
    }
]
```

```python
ENABLE_THINKING = True
MAX_NEW_TOKENS = 3000


inputs = tokenizer.apply_chat_template(
    messages,
    add_generation_prompt=True,
    tokenize=True,
    return_dict=True,
    return_tensors="pt",
    enable_thinking=ENABLE_THINKING,
).to(model.device)


if ENABLE_THINKING:
    generation_kwargs = {
        "max_new_tokens": MAX_NEW_TOKENS,
        "do_sample": True,
        "temperature": 1.0,
        "top_p": 0.95,
        "top_k": 20,
    }
else:
    generation_kwargs = {
        "max_new_tokens": MAX_NEW_TOKENS,
        "do_sample": True,
        "temperature": 0.7,
        "top_p": 0.8,
        "top_k": 20,
    }


with torch.inference_mode():
    outputs = model.generate(
        **inputs,
        **generation_kwargs,
    )


generated_ids = outputs[0][
    inputs["input_ids"].shape[-1]:
]

response = tokenizer.decode(
    generated_ids,
    skip_special_tokens=True,
).strip()


# Remove Qwen thinking trace if any before parsing JSON
if "</think>" in response:
    final_response = response.split("</think>", 1)[1].strip()
else:
    final_response = response


print(final_response)


parsed_response = json.loads(final_response)


output_path.parent.mkdir(
    parents=True,
    exist_ok=True,
)

output_path.write_text(
    json.dumps(
        parsed_response,
        indent=2,
        ensure_ascii=False,
    ),
    encoding="utf-8",
)

print(f"Specification saved to: {output_path}")
```
